Assignment 26: Combine Normal Arguments and *args
Write a function named student_marks().

Requirements
The first parameter should be the student name.
Remaining positional arguments should represent marks.
Calculate total and average marks.
Return both total and average.
Example Call
student_marks("Ravi", 80, 75, 90, 88)

```python
def student_marks(name, *marks):
    total = 0


    for mark in marks:
        total += mark


    average = total / len(marks)

    return total, average

total, average = student_marks("Ravi", 80, 75, 90, 88)

print("Student Name :", "Ravi")
print("Total Marks  :", total)
print("Average Marks:", average)
```

Study this general order:

def function(positional_only, /, normal, default=value, *args, keyword_only, **kwargs):


```python
def generate_report(report_id, /, title, author="Unknown", *sections, status, **details):
    print("Report Details")
    print("----------------------")
    print("Report ID :", report_id)      # Positional-only parameter
    print("Title     :", title)          # Normal parameter
    print("Author    :", author)         # Default parameter
    print("Status    :", status)         # Keyword-only parameter

    print("\nSections:")
    if sections:
        for section in sections:
            print("-", section)
    else:
        print("No sections provided.")

    print("\nAdditional Details:")
    if details:
        for key, value in details.items():
            print(f"{key} : {value}")
    else:
        print("No additional details.")



generate_report(
    101,
    "Annual Sales Report",
    "Ravi Kumar",
    "Introduction",
    "Sales Analysis",
    "Conclusion",
    status="Completed",
    department="Sales",
    year=2025,
    location="Hyderabad"
)
```

Assignment 31: Return Multiple Values
Write a function named number_statistics().

Requirements
Accept a list of numbers.
Return minimum, maximum, total, and average.
Use tuple unpacking to store returned values.
You may use built-in functions in this assignment.


```python
def number_statistics(numbers):
    minimum = min(numbers)
    maximum = max(numbers)
    total = sum(numbers)
    average = total / len(numbers)

    # Return multiple values as a tuple
    return minimum, maximum, total, average


# List of numbers
numbers = [10, 20, 30, 40, 50]

# Tuple unpacking to store returned values
minimum, maximum, total, average = number_statistics(numbers)

# Display results
print("Numbers :", numbers)
print("Minimum :", minimum)
print("Maximum :", maximum)
print("Total   :", total)
print("Average :", average)
```

Assignment 32: Function Returning a Dictionary
Write a function named analyze_text().

Requirements
Accept a string.
Return a dictionary containing:
Total characters
Total words
Total vowels
Total digits
Total spaces
Display every result clearly.

```python
def analyze_text(text):
    vowels = "aeiouAEIOU"

    total_characters = len(text)
    total_words = len(text.split())
    total_vowels = 0
    total_digits = 0
    total_spaces = 0


    for char in text:
        if char in vowels:
            total_vowels += 1
        elif char.isdigit():
            total_digits += 1
        elif char == " ":
            total_spaces += 1


    return {
        "Total Characters": total_characters,
        "Total Words": total_words,
        "Total Vowels": total_vowels,
        "Total Digits": total_digits,
        "Total Spaces": total_spaces
    }
text = "Python Programming 123"

result = analyze_text(text)

print("Text Analysis Report")
print("--------------------")

for key, value in result.items():
    print(f"{key}: {value}")
```

Assignment 33: Function Returning a List
Write a function named get_even_numbers().

Requirements
Accept a starting number and ending number.
Return a list containing all even numbers in the range.
Validate that the starting number is smaller than the ending number.


```python
def get_even_numbers(start, end):

    if start >= end:
        return "Invalid range: Starting number must be smaller than ending number."

    even_numbers = []


    for num in range(start, end + 1):
        if num % 2 == 0:
            even_numbers.append(num)


    return even_numbers

result = get_even_numbers(1, 20)

print("Even Numbers:", result)


invalid_result = get_even_numbers(20, 10)

print(invalid_result)
```

Assignment 34: Prime Number Function
Write a function named is_prime().

Requirements
Accept one integer.
Return True when the number is prime.
Return False otherwise.
Use the function to print all prime numbers from 1 to 100.
Hint
A prime number has exactly two factors.

```python
def is_prime(number):

    if number < 2:
        return False


    for i in range(2, number):
        if number % i == 0:
            return False

    return True

print("Prime numbers from 1 to 100:")

for num in range(1, 101):
    if is_prime(num):
        print(num, end=" ")
```

Assignment 35: Fibonacci Function
Write a function named generate_fibonacci().

Requirements
Accept the number of terms.
Return the complete Fibonacci series as a list.
Handle zero and negative term counts.
Do not use recursion in this assignment.
Example Input
Enter number of terms: 7
Expected Output
[0, 1, 1, 2, 3, 5, 8]


```python
def generate_fibonacci(terms):

    if terms <= 0:
        return []

    fibonacci = []

    a = 0
    b = 1


    for i in range(terms):
        fibonacci.append(a)
        a, b = b, a + b

    return fibonacci

terms = int(input("Enter number of terms: "))
result = generate_fibonacci(terms)


print(result)
```

Assignment 41: Sort Tuples Using Lambda
Given the following list:

students = [
    ("Ravi", 82),
    ("Anitha", 95),
    ("Kiran", 76),
    ("Meena", 89)
]
Requirements
Sort students by marks in ascending order.
Sort students by marks in descending order.
Use sorted() with a lambda as the key.
Print both results.

```python
students = [
    ("Ravi", 82),
    ("Anitha", 95),
    ("Kiran", 76),
    ("Meena", 89)
]


ascending_order = sorted(students, key=lambda student: student[1])


descending_order = sorted(students, key=lambda student: student[1], reverse=True)


print("Students sorted by marks (Ascending):")
print(ascending_order)

print("\nStudents sorted by marks (Descending):")
print(descending_order)
```

Assignment 45: Lambda Discount Calculator
Create a lambda expression that accepts product price and discount percentage.

Requirements
Calculate the final price after discount.
Validate inputs before calling the lambda.
Display the discount amount and final price.
Formula
Final Price = Price - (Price × Discount / 100)

```python
price = float(input("Enter product price: "))
discount = float(input("Enter discount percentage: "))

if price < 0:
    print("Invalid price. Price cannot be negative.")

elif discount < 0 or discount > 100:
    print("Invalid discount percentage. Enter a value between 0 and 100.")

else:

    calculate_price = lambda p, d: p - (p * d / 100)


    discount_amount = price * discount / 100

    final_price = calculate_price(price, discount)


    print("Original Price :", price)
    print("Discount Amount:", discount_amount)
    print("Final Price    :", final_price)
```
