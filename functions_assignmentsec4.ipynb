Assignment 46: Square Numbers Using map()
Given a list of numbers, generate their squares using map().

Requirements
Accept numbers from the user.
Use map() with a lambda expression.
Convert the map object into a list.
Print original and squared lists.
Do not use a loop or list comprehension for transformation.

```python
numbers = list(map(int, input("Enter numbers separated by spaces: ").split()))

squared_numbers = list(map(lambda x: x ** 2, numbers))

print("Original List:", numbers)
print("Squared List:", squared_numbers)
```

Assignment 51: Add Two Lists Using map()
Given two lists of equal length, add corresponding values.

Requirements
Use map() with two iterables.
Do not use an index-based loop.
Validate that both lists contain the same number of elements.
Example
List 1: [10, 20, 30]
List 2: [1, 2, 3]
Result: [11, 22, 33]

```python
list1 = list(map(int, input("Enter the first list of numbers: ").split()))
list2 = list(map(int, input("Enter the second list of numbers: ").split()))

if len(list1) != len(list2):
    print("Error: Both lists must contain the same number of elements.")
else:

    result = list(map(lambda x, y: x + y, list1, list2))

    print("List 1:", list1)
    print("List 2:", list2)
    print("Result:", result)
```

Assignment 52: Multiply Three Lists
Create three numerical lists of equal length.

Requirements
Multiply corresponding values from all three lists.
Use map() with a lambda that accepts three parameters.
Print the resulting list.
Example
[1, 2, 3]
[4, 5, 6]
[2, 2, 2]

Result: [8, 20, 36]

```python
list1 = list(map(int, input("Enter the first list of numbers: ").split()))
list2 = list(map(int, input("Enter the second list of numbers: ").split()))
list3 = list(map(int, input("Enter the third list of numbers: ").split()))

if len(list1) == len(list2) == len(list3):

    result = list(map(lambda x, y, z: x * y * z, list1, list2, list3))


    print("Result:", result)
else:
    print("Error: All three lists must contain the same number of elements.")
```

Assignment 54: Calculate Final Product Prices
Given a list of product prices, add 18% tax to every price.

Requirements
Use map().
Return prices rounded to two decimal places.
Display original price and final price together.
Use zip() for the final display.
Formula
Final Price = Price + (Price × 18 / 100)

```python
prices = list(map(float, input("Enter product prices separated by spaces: ").split()))


final_prices = list(map(lambda price: round(price + (price * 18 / 100), 2), prices))
print("\nOriginal Price -> Final Price")
for original, final in zip(prices, final_prices):
    print(f"{original:.2f} -> {final:.2f}")
```

Assignment 56: Mask Email Addresses
Given a list of email addresses, mask the username portion.

Example
bharath@gmail.com → b******@gmail.com
Requirements
Use map().
Keep the first character of the username.
Replace remaining username characters with *.
Keep the domain unchanged.
Create a separate named function for masking.

```python
def mask_email(email):
    username, domain = email.split("@")
    masked_username = username[0] + "*" * (len(username) - 1)
    return masked_username + "@" + domain

emails = input("Enter email addresses separated by spaces: ").split()

masked_emails = list(map(mask_email, emails))

print("Masked Emails:", masked_emails)
```

Assignment 57: Format Phone Numbers
Given a list of 10-digit phone numbers as strings, format each one.

Expected Format
9876543210 → 987-654-3210
Requirements
Use map().
Validate the length and digits.
Return "Invalid" for invalid phone numbers.

```python
def format_phone(number):
    if len(number) == 10 and number.isdigit():
        return number[:3] + "-" + number[3:6] + "-" + number[6:]
    else:
        return "Invalid"

phone_numbers = input("Enter 10-digit phone numbers separated by spaces: ").split()


formatted_numbers = list(map(format_phone, phone_numbers))

print("Formatted Phone Numbers:", formatted_numbers)
```

Assignment 59: Calculate Employee Bonuses
Given employee salaries, calculate a 10% bonus for each employee.

Requirements
Use map().
Return tuples containing salary, bonus, and total salary.
Round all decimal values to two places.
Expected Tuple Format
(50000, 5000, 55000)


```python
salaries = list(map(float, input("Enter employee salaries separated by spaces: ").split()))
employee_bonus = list(map(lambda salary: (
    round(salary, 2),
    round(salary * 10 / 100, 2),
    round(salary + (salary * 10 / 100), 2)
), salaries))


print("Salary, Bonus, Total Salary:")
for record in employee_bonus:
    print(record)
```

Assignment 60: Apply Different Operations Using Map
Given a list of numbers, generate:

Squares
Cubes
Doubles
Half values
Requirements
Use four separate map() operations.
Use lambda expressions.
Print every resulting list with a clear heading.

```python
numbers = list(map(float, input("Enter numbers separated by spaces: ").split()))

squares = list(map(lambda x: x ** 2, numbers))
cubes = list(map(lambda x: x ** 3, numbers))
doubles = list(map(lambda x: x * 2, numbers))
half_values = list(map(lambda x: x / 2, numbers))

print("Original List:", numbers)
print("Squares:", squares)
print("Cubes:", cubes)
print("Doubles:", doubles)
print("Half Values:", half_values)
```
