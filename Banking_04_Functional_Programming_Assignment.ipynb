### Q1. Write a function for line total for banking case 1; use sample reference `ACC-1001` and explain the result.

**Hints**

1. First express the input and return value of one pure function.
2. Use `filter` for selection, `map` for transformation, and `reduce` only when a clearer built-in is not available.
3. Test the pipeline with an empty collection and avoid mutating account records in place.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python

from functools import reduce

reference = "ACC-1001"

def line_total(items):
    if not isinstance(items, list):
        return 0

    valid_items = filter(
        lambda x: isinstance(x, dict)
        and isinstance(x.get("qty"), (int, float))
        and isinstance(x.get("price"), (int, float))
        and x["qty"] >= 0 and x["price"] >= 0,
        items
    )

    totals = map(lambda x: x["qty"] * x["price"], valid_items)

    return sum(totals)

# Valid
items = [
    {"name": "Deposit", "qty": 2, "price": 500},
    {"name": "Service", "qty": 1, "price": 100}
]

# Empty input
empty_items = []

print("Reference:", reference)
print("Line Total:", line_total(items))
print("Empty Total:", line_total(empty_items))
print("Invalid Input:", line_total("invalid"))
```

### Q2. Return a discounted amount for banking case 1; use sample reference `ACC-1002` and explain the result.

**Hints**

1. First express the input and return value of one pure function.
2. Use `filter` for selection, `map` for transformation, and `reduce` only when a clearer built-in is not available.
3. Test the pipeline with an empty collection and avoid mutating account records in place.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python


reference = "ACC-1002"

def discounted_amount(items, discount_rate):
    if not isinstance(items, list) or not isinstance(discount_rate, (int, float)):
        return 0

    if not 0 <= discount_rate <= 100:
        return 0

    valid_items = filter(
        lambda x: isinstance(x, (int, float)) and x >= 0,
        items
    )

    discounted = map(
        lambda x: x * (1 - discount_rate / 100),
        valid_items
    )

    return sum(discounted)

# Valid
items = [1000, 500, 200]

# Empty input
empty_items = []

print("Reference:", reference)
print("Discounted Amount:", discounted_amount(items, 10))
print("Empty Total:", discounted_amount(empty_items, 10))
print("Invalid Input:", discounted_amount("invalid", 10))
print("Boundary (0%):", discounted_amount([1000], 0))
```

### Q13. Reduce a basket to its total for banking case 1; use sample reference `ACC-1013` and explain the result.

**Hints**

1. First express the input and return value of one pure function.
2. Use `filter` for selection, `map` for transformation, and `reduce` only when a clearer built-in is not available.
3. Test the pipeline with an empty collection and avoid mutating account records in place.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python

from functools import reduce

reference = "ACC-1013"

def basket_total(items):
    if not isinstance(items, list):
        return 0

    valid_items = filter(
        lambda x: isinstance(x, (int, float)) and x >= 0,
        items
    )

    return reduce(lambda a, b: a + b, valid_items, 0)

items = [100, 200, 300]

print("Reference:", reference)
print("Basket Total:", basket_total(items))
print("Empty:", basket_total([]))
print("Invalid:", basket_total("invalid"))
print("Boundary:", basket_total([0, 100]))
```

```python

reference = "ACC-1098"

def retry(times):
    def decorator(func):
        def wrapper(amount):
            for attempt in range(times):
                try:
                    return func(amount)
                except Exception:
                    if attempt == times - 1:
                        return 0
            return 0
        return wrapper
    return decorator


@retry(3)
def transfer(amount):
    if not isinstance(amount, (int, float)) or amount < 0:
        raise ValueError("Invalid transfer amount")
    return amount


amounts = [1000, 500, 200]

valid_amounts = filter(
    lambda x: isinstance(x, (int, float)) and x >= 0,
    amounts
)

result = list(map(transfer, valid_amounts))

print(reference)
print(result)

# Edge cases
print(list(map(transfer, [])))
print(transfer("1000"))
print(transfer(0))
print(transfer(-500))
```

### Q100. Design an end-to-end transactions analytics pipeline for banking case 3; use sample reference `ACC-1100` and explain the result.

2. Use `filter` for selection, `map` for transformation, and `reduce` only when a clearer built-in is not available.
3. Test the pipeline with an empty collection and avoid mutating account records in place.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python

from functools import reduce

reference = "ACC-1100"

def transactions_analytics(transactions):
    if not isinstance(transactions, list):
        return {
            "count": 0,
            "total": 0,
            "average": 0
        }

    # Filter valid transactions
    valid = list(filter(
        lambda x: isinstance(x, (int, float)) and x >= 0,
        transactions
    ))

    # Apply 18% tax
    taxed = list(map(lambda x: x * 1.18, valid))

    # Calculate total
    total = reduce(lambda a, b: a + b, taxed, 0)

    count = len(taxed)
    average = total / count if count else 0

    return {
        "count": count,
        "total": total,
        "average": average
    }


transactions = [1000, 500, 200]

result = transactions_analytics(transactions)

print(reference)
print(result)

# Edge cases
print(transactions_analytics([]))
print(transactions_analytics([100, "x", 200]))
print(transactions_analytics([0]))
print(transactions_analytics([-500]))
print(transactions_analytics("1000"))
```
