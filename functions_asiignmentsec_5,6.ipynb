Assignment 67: Filter Prime Numbers
Accept a list of integers.

Requirements
Write a reusable is_prime() function.
Pass the function to filter().
Print only prime values.
Handle values below 2 correctly.


```python
def is_prime(num):
    if num < 2:
        return False
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            return False
    return True

numbers = list(map(int, input("Enter integers separated by spaces: ").split()))

prime_numbers = list(filter(is_prime, numbers))

print("Prime numbers:", prime_numbers)
```

Assignment 68: Filter Palindrome Words
Accept a sentence from the user.

Requirements
Split it into words.
Use filter() to find palindrome words.
Ignore case and punctuation.
Write a separate function for palindrome checking.

```python
def is_palindrome(word):

    cleaned = word.strip(string.punctuation).lower()
    return cleaned != "" and cleaned == cleaned[::-1]

sentence = input("Enter a sentence: ")
words = sentence.split()

palindrome_words = list(filter(is_palindrome, words))

print("Palindrome words:", palindrome_words)
```

Assignment 69: Filter Valid Email Addresses
Given a list of email strings:

Requirements
Use filter().
A basic valid email should:
Contain exactly one @.
Have characters before @.
Contain a dot after @.
Return only emails satisfying these conditions.
Note
A complete regular expression is not required.

```python
def is_valid_email(email):
    if email.count("@") != 1:
        return False

    username, domain = email.split("@")

    return (
        username != "" and
        "." in domain and
        domain.index(".") > 0 and
        domain[-1] != "."
    )

emails = [
    "alice@example.com",
    "bob@gmail",
    "@yahoo.com",
    "john.doe@company.org",
    "test@@mail.com",
    "user@mail.",
    "hello@domain.co.in",
    "invalid.com"
]

valid_emails = list(filter(is_valid_email, emails))

print("Valid email addresses:", valid_emails)
```

Assignment 71: Filter High-Salary Employees
Create a list of employee dictionaries.

Requirements
Ask the user for a salary threshold.
Use filter() to select employees with salaries greater than or equal to the threshold.
Sort selected employees by salary in descending order.
Use a lambda for both filtering and sorting.

```python
employees = [
    {"name": "Alice", "salary": 50000},
    {"name": "Bob", "salary": 75000},
    {"name": "Charlie", "salary": 60000},
    {"name": "David", "salary": 90000},
    {"name": "Eva", "salary": 45000}
]

threshold = int(input("Enter salary threshold: "))

high_salary_employees = list(
    filter(lambda emp: emp["salary"] >= threshold, employees)
)

high_salary_employees.sort(
    key=lambda emp: emp["salary"],
    reverse=True
)

print("\nEmployees with high salaries:")
for emp in high_salary_employees:
    print(f"Name: {emp['name']}, Salary: {emp['salary']}")
```

Assignment 72: Filter Products in Stock
Given product dictionaries containing name, price, and stock:

Requirements
Use filter() to return products with stock greater than zero.
Create another result containing products priced below a user-defined budget.
Create a final result satisfying both conditions.

```python
products = [
    {"name": "Laptop", "price": 60000, "stock": 5},
    {"name": "Phone", "price": 30000, "stock": 0},
    {"name": "Tablet", "price": 25000, "stock": 10},
    {"name": "Headphones", "price": 5000, "stock": 20},
    {"name": "Keyboard", "price": 3000, "stock": 0}
]

in_stock_products = list(
    filter(lambda product: product["stock"] > 0, products)
)

budget = int(input("Enter your budget: "))

budget_products = list(
    filter(lambda product: product["price"] < budget, products)
)

available_budget_products = list(
    filter(lambda product: product["stock"] > 0 and product["price"] < budget, products)
)

print("\nProducts in stock:")
for product in in_stock_products:
    print(product)

print("\nProducts below budget:")
for product in budget_products:
    print(product)

print("\nProducts in stock and below budget:")
for product in available_budget_products:
    print(product)
```

 Assignment 74: Filter Perfect Squares
Accept a list of positive integers.

Requirements
Use filter() to return values that are perfect squares.
Write a named function for checking perfect squares.
Use the math module if required.
Example
Input: [1, 2, 4, 8, 9, 15, 16]
Output: [1, 4, 9, 16]

```python
import math

def is_perfect_square(num):
    if num <= 0:
        return False

    root = math.isqrt(num)
    return root * root == num

numbers = list(map(int, input("Enter positive integers separated by spaces: ").split()))

perfect_squares = list(filter(is_perfect_square, numbers))

print("Perfect squares:", perfect_squares)
```

Assignment 75: Filter Dates by Year
Given dates in DD-MM-YYYY format:

Requirements
Ask the user for a year.
Use filter() to return dates belonging to that year.
Handle invalid date strings safely.
Use datetime.strptime() for validation.

```python
from datetime import datetime

dates = [
    "15-08-2024",
    "01-01-2025",
    "28-02-2025",
    "31-12-2023",
    "invalid-date",
    "30-02-2025"
]

def is_year_match(date_string):
    try:
        date = datetime.strptime(date_string, "%d-%m-%Y")
        return date.year == year
    except ValueError:
        return False
year = int(input("Enter year: "))

filtered_dates = list(filter(is_year_match, dates))

print("Dates in", year, ":")
print(filtered_dates)
```

Assignment 81: Calculate Factorial Using reduce()
Write a function that calculates factorial using reduce().

Requirements
Accept a non-negative integer.
Generate numbers from 1 to the given number.
Multiply them using reduce().
Return 1 when the input is 0.
Reject negative values.

```python
def factorial(n):

    if n < 0:
        return "Factorial is not defined for negative numbers"

    if n == 0:
        return 1

    numbers = range(1, n + 1)
    result = reduce(lambda accumulator, value: accumulator * value, numbers)

    return result

num = int(input("Enter a non-negative integer: "))

print("Factorial:", factorial(num))
```

Assignment 82: Calculate Total Cart Price
Given a shopping cart:

cart = [
    {"product": "Laptop", "price": 70000, "quantity": 1},
    {"product": "Mouse", "price": 800, "quantity": 2},
    {"product": "Keyboard", "price": 1500, "quantity": 1}
]
Requirements
Use map() to calculate each item's total.
Use reduce() to calculate the grand total.
Print item-wise totals and the grand total.

```python
cart = [
    {"product": "Laptop", "price": 70000, "quantity": 1},
    {"product": "Mouse", "price": 800, "quantity": 2},
    {"product": "Keyboard", "price": 1500, "quantity": 1}
]

item_totals = list(map(
    lambda item: {
        "product": item["product"],
        "total": item["price"] * item["quantity"]
    },
    cart
))

grand_total = reduce(
    lambda accumulator, item: accumulator + item["total"],
    item_totals,
    0
)

print("Item-wise totals:")
for item in item_totals:
    print(f"{item['product']}: {item['total']}")

# Print grand total
print("Grand Total:", grand_total)
```

```python
sentence = input("Enter a sentence: ")

words = sentence.split()

if words:
    longest_word = reduce(
        lambda accumulator, word: accumulator if len(accumulator) >= len(word) else word,
        words
    )
    print("Longest word:", longest_word)
else:
    print("The sentence is empty.")
```

Assignment 85: Compose Map, Filter and Reduce
Accept a list of integers.

Requirements
Perform the following operations:

Use filter() to select even numbers.
Use map() to square the selected numbers.
Use reduce() to calculate the sum of the squares.
Example
Input: [1, 2, 3, 4, 5, 6]
Even numbers: [2, 4, 6]
Squares: [4, 16, 36]
Final total: 56

```python
numbers = list(map(int, input("Enter integers separated by spaces: ").split()))

even_numbers = list(filter(lambda x: x % 2 == 0, numbers))

squares = list(map(lambda x: x ** 2, even_numbers))

total = reduce(lambda accumulator, value: accumulator + value, squares, 0)

print("Even numbers:", even_numbers)
print("Squares:", squares)
print("Final total:", total)
```
