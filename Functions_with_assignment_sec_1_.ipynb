Create four separate functions:

Requirements

Print the calculated result.
Example Input
Enter first number: 12
Enter second number: 4

Expected Output

```python

def add(a, b):
    return a + b
def subtract(a, b):
    return a - b
def multiply(a, b):
    return a * b
def divide(a, b):
    if b == 0:
        return "Error! Division by zero is not allowed."
    return a / b
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

operation = input("Enter operation (add, subtract, multiply, divide): ").lower()

if operation == "add":
    result = add(num1, num2)
elif operation == "subtract":
    result = subtract(num1, num2)
elif operation == "multiply":
    result = multiply(num1, num2)
elif operation == "divide":
    result = divide(num1, num2)
else:
    result = "Invalid operation!"

print("Result:", result)
```

Assignment 5: Find the Largest Number
Write a function named find_largest() that accepts three numbers.

Requirements
Do not use the built-in max() function.
Compare the numbers using conditional statements.
Return the largest number.
Accept three values from the user.
Example Input
Enter three numbers: 25 60 42
Expected Output
Largest number: 60
Hint
Start by assuming the first number is the largest.

```python
def find_largest(num1, num2, num3):
    largest = num1

    if num2 > largest:
        largest = num2

    if num3 > largest:
        largest = num3

    return largest


num1, num2, num3 = map(int, input("Enter three numbers: ").split())

result = find_largest(num1, num2, num3)
print("Largest number:", result)
```

Assignment 8: Area and Circumference of a Circle
Create two functions:

circle_area(radius)
circle_circumference(radius)
Requirements
Accept the radius from the user.
Use math.pi.
Return the area and circumference separately.
Display both values rounded to two decimal places.
Formulas
Area = π × radius²
Circumference = 2 × π × radius
Hint
Import the math module.

```python
import math

def circle_area(radius):
    return math.pi * (radius ** 2)
def circle_circumference(radius):
    return 2 * math.pi * radius

radius = float(input("Enter the radius: "))
area = circle_area(radius)
circumference = circle_circumference(radius)

print("Area:", round(area, 2))
print("Circumference:", round(circumference, 2))
```

Assignment 9: Student Grade Calculator
Write a function named calculate_grade().

Requirements
Accept a student's marks.
Return the grade according to the following rules:
90–100: A
80–89: B
70–79: C
60–69: D
Below 60: F
Validate that marks are between 0 and 100.
Example Input
Enter marks: 84
Expected Output
Grade: B


```python
def calculate_grade(marks):
    if marks < 0 or marks > 100:
        return "Invalid marks! Please enter marks between 0 and 100."
    elif marks >= 90:
        return "A"
    elif marks >= 80:
        return "B"
    elif marks >= 70:
        return "C"
    elif marks >= 60:
        return "D"
    else:
        return "F"

marks = float(input("Enter marks: "))

grade = calculate_grade(marks)

if grade in ["A", "B", "C", "D", "F"]:
    print("Grade:", grade)
else:
    print(grade)
```

Assignment 12: Check Leap Year
Write a function named is_leap_year().

Requirements
Accept a year as an integer.
Return True when it is a leap year.
Otherwise, return False.
Print an appropriate message.
Rules
A leap year is:

Divisible by 400, or
Divisible by 4 but not divisible by 100.


```python
def is_leap_year(year):
    if year % 400 == 0:
        return True
    elif year % 4 == 0 and year % 100 != 0:
        return True
    else:
        return False

year = int(input("Enter a year: "))

if is_leap_year(year):
    print(year, "is a leap year.")
else:
    print(year, "is not a leap year.")
```

Assignment 13: Factorial Using a Function
Write a function named calculate_factorial().

Requirements
Accept a non-negative integer.
Calculate its factorial using a loop.
Do not use math.factorial().
Return the factorial.
Handle negative input.
Example Input
Enter a number: 5
Expected Output
Factorial: 120

```python
def calculate_factorial(number):
    if number < 0:
        return "Error! Factorial is not defined for negative numbers."

    factorial = 1
    for i in range(1, number + 1):
        factorial *= i

    return factorial

num = int(input("Enter a number: "))

result = calculate_factorial(num)

print("Factorial:", result)
```

Assignment 16: Reverse a String
Write a function named reverse_string().

Requirements
Accept a string.
Return its reversed version.
Solve it first using slicing.
Write another function that reverses it using a loop.
Example Input
Enter text: Python
Expected Output
Reversed text: nohtyP

```python
def reverse_string(text):
    return text[::-1]

def reverse_string_loop(text):
    reversed_text = ""

    for char in text:
        reversed_text = char + reversed_text

    return reversed_text

text = input("Enter text: ")
reversed_text = reverse_string(text)
reversed_text_loop = reverse_string_loop(text)

print("Reversed text:", reversed_text)
print("Reversed text using loop:", reversed_text_loop)
```

Assignment 17: Palindrome Checker
Write a function named is_palindrome().

Requirements
Accept a string.
Ignore spaces and letter case.
Return True when the string is a palindrome.
Print a meaningful result.
Example Input
Enter text: Never odd or even
Expected Output
The text is a palindrome.
Hint
Clean the string before comparing it with its reverse.

```python
def is_palindrome(text):

    cleaned_text = text.replace(" ", "").lower()

    return cleaned_text == cleaned_text[::-1]

text = input("Enter text: ")

if is_palindrome(text):
    print("The text is a palindrome.")
else:
    print("The text is not a palindrome.")
```

Assignment 19: Find Maximum in a List
Write a function named find_list_maximum().

Requirements
Accept a list of numbers.
Do not use max().
Return the largest list element.
Handle an empty list by returning an appropriate message.

```python
def find_list_maximum(numbers):
    if len(numbers) == 0:
        return "The list is empty."

    maximum = numbers[0]

    for number in numbers:
        if number > maximum:
            maximum = number

    return maximum

# Accept numbers from the user
numbers = list(map(int, input("Enter numbers: ").split()))

# Call the function and print the result
result = find_list_maximum(numbers)

print("Maximum number:", result)
```
