### Q1. Find a account code for banking case 1; use sample reference `ACC-1001` and explain the result.

```python

import re

ACCOUNT_CODE_RE = re.compile(r"(?P<prefix>ACC)-(?P<number>\d{4})")

def validate_account_code(value):
    if not isinstance(value, str):
        return False
    return ACCOUNT_CODE_RE.fullmatch(value) is not None


# Sample
account_code = "ACC-1001"

match = ACCOUNT_CODE_RE.fullmatch(account_code)

if match:
    print("Valid account code")
    print("Prefix:", match.group("prefix"))
    print("Account number:", match.group("number"))
else:
    print("Invalid account code")
```

```python

import re

AMOUNT_RE = re.compile(
    r"(?P<amount>₹\d{1,3}(?:,\d{3})*(?:\.\d{2})?|₹\d+(?:\.\d{2})?)"
)

def extract_amounts(value):
    if not isinstance(value, str) or not value:
        return []

    return [
        match.group("amount")
        for match in AMOUNT_RE.finditer(value)
    ]


# Sample reference
account_code = "ACC-1003"
text = "Account ACC-1003 received ₹1,500.00 and paid ₹250.00. Balance: ₹10,000.00."

amounts = extract_amounts(text)

print("Account:", account_code)
print("Amounts:", amounts)
```

### Q4. Match a receipt date for banking case 1; use sample reference `ACC-1004` and explain the result.

**Hints**

1. Write representative valid and invalid samples before writing the pattern.
2. Use raw strings, named groups for extraction, and `fullmatch` for whole-field validation.
3. Never treat regex alone as authorization or full semantic validation; verify dates and checksums separately.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python

import re
from datetime import datetime

DATE_RE = re.compile(
    r"(?P<day>\d{2})/(?P<month>\d{2})/(?P<year>\d{4})"
)

def validate_receipt_date(value):
    if not isinstance(value, str):
        return False

    match = DATE_RE.fullmatch(value)
    if not match:
        return False

    try:
        datetime.strptime(value, "%d/%m/%Y")
        return True
    except ValueError:
        return False


# Sample reference
account_code = "ACC-1004"
receipt_date = "15/08/2026"

if validate_receipt_date(receipt_date):
    match = DATE_RE.fullmatch(receipt_date)
    print("Valid receipt date")
    print("Account:", account_code)
    print("Day:", match.group("day"))
    print("Month:", match.group("month"))
    print("Year:", match.group("year"))
else:
    print("Invalid receipt date")
```

### Q6. Remove extra spaces for banking case 1; use sample reference `ACC-1006` and explain the result.

**Hints**

1. Write representative valid and invalid samples before writing the pattern.
2. Use raw strings, named groups for extraction, and `fullmatch` for whole-field validation.
3. Never treat regex alone as authorization or full semantic validation; verify dates and checksums separately.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python

import re

SPACE_RE = re.compile(r"(?P<text>\S(?:.*?\S)?)")

def remove_extra_spaces(value):
    if not isinstance(value, str):
        return None

    match = SPACE_RE.fullmatch(value.strip())

    if not match:
        return ""

    return re.sub(r"\s+", " ", match.group("text"))


# Sample reference
account_reference = "  ACC-1006   "

result = remove_extra_spaces(account_reference)

print("Original:", repr(account_reference))
print("Result:", result)
```

```python
import re

# Valid:
# ACC-1007
# ACC-1007,LOAN-204
# ACC-1007,LOAN-204,CARD-3001

# Invalid:
# ""
# ACC-1007,
# ,ACC-1007
# ACC-1007,,LOAN-204
# ACC-10A7

TAG_LIST_RE = re.compile(
    r"(?P<tags>ACC-\d{4}(?:,(?:ACC-\d{4}|LOAN-\d{3}|CARD-\d{4}))*)"
)

def split_tags(value):
    if not isinstance(value, str):
        raise TypeError("tag list must be a string")

    if not TAG_LIST_RE.fullmatch(value):
        raise ValueError("invalid banking tag list")

    return value.split(",")


# Sample reference
sample = "ACC-1007"
print(split_tags(sample))
# ['ACC-1007']

# Multiple tags
print(split_tags("ACC-1007,LOAN-204,CARD-3001"))
# ['ACC-1007', 'LOAN-204', 'CARD-3001']

# Realistic failure path
try:
    split_tags("ACC-1007,,LOAN-204")
except ValueError as e:
    print(e)
# invalid banking tag list

# Boundary / invalid-type tests
for value in ["", "ACC-1007,", ",ACC-1007", "ACC-10A7", None, 1007]:
    try:
        print(value, "->", split_tags(value))
    except (TypeError, ValueError) as e:
        print(value, "->", type(e).__name__, e)

# Note:
# Regex validates format only. In a banking system, separately verify
# account existence, dates, checksums, authorization, and other semantics.
```

### Q19. Match a receipt date for banking case 2; use sample reference `ACC-1019` and explain the result.

**Hints**

1. Write representative valid and invalid samples before writing the pattern.
2. Use raw strings, named groups for extraction, and `fullmatch` for whole-field validation.
3. Never treat regex alone as authorization or full semantic validation; verify dates and checksums separately.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python

import re
from datetime import date

pattern = re.compile(
    r"(?P<year>\d{4})-(?P<month>\d{2})-(?P<day>\d{2})"
)

def match_receipt_date(value):
    if not isinstance(value, str):
        return None

    match = pattern.fullmatch(value)
    if not match:
        return None

    parts = {name: int(value) for name, value in match.groupdict().items()}

    # Semantic date validation
    try:
        return date(**parts)
    except ValueError:
        return None


result = match_receipt_date("2026-09-11")
```

### Q80. Create a regex-based data quality report for banking case 1; use sample reference `ACC-1080` and explain the result.

### Q100. Create a regex-based data quality report for banking case 3; use sample reference `ACC-1100` and explain the result.

**Hints**

1. Write representative valid and invalid samples before writing the pattern.
2. Use raw strings, named groups for extraction, and `fullmatch` for whole-field validation.
3. Never treat regex alone as authorization or full semantic validation; verify dates and checksums separately.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python

import re

reference = "ACC-1100"

account_pattern = r"(?P<account>ACC-\d{4})"
amount_pattern = r"(?P<currency>USD|INR|EUR|GBP)\s+(?P<amount>\d{1,10}\.\d{2})"

def data_quality_report(account, amount):
    if not isinstance(account, str) or not account:
        account_result = "Invalid"
    else:
        m = re.fullmatch(account_pattern, account)
        account_result = m.groupdict() if m else "Invalid"

    if not isinstance(amount, str) or not amount:
        amount_result = "Invalid"
    else:
        m = re.fullmatch(amount_pattern, amount)
        amount_result = m.groupdict() if m else "Invalid"

    status = "PASS" if isinstance(account_result, dict) and isinstance(amount_result, dict) else "FAIL"

    return {
        "account": account_result,
        "amount": amount_result,
        "status": status
    }

print("Reference:", reference)
print(data_quality_report("ACC-1100", "INR 500.00"))
print(data_quality_report("ACC-110", "INR 500.00"))
print(data_quality_report("ACC-1100", "USD 10.555"))
print(data_quality_report("", "USD 100.00"))
print(data_quality_report(None, "USD 100.00"))
```
