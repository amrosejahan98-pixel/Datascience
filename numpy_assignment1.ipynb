Question 36: Store-Wise Profit Margin
Calculate the overall profit margin percentage for each store using:

Store Profit Margin =
Total Store Profit / Total Store Sales × 100

```python
import numpy as np

sales_data = np.array([
    [200, 150, 180, 220],
    [250, 170, 200, 240],
    [180, 160, 190, 210],
    [300, 200, 220, 260]
])

cost_data = np.array([
    [80, 60, 50, 70],
    [90, 65, 55, 75],
    [85, 70, 60, 80],
    [95, 75, 65, 85]
])

stores = np.array([
    "Bengaluru",
    "Hyderabad",
    "Chennai",
    "Mumbai"
])

profit_data = sales_data - cost_data

total_store_sales = np.sum(sales_data, axis=1)
total_store_profit = np.sum(profit_data, axis=1)

store_profit_margin = (total_store_profit / total_store_sales) * 100

store_profit_margin = np.round(store_profit_margin, 2)

print("Store-Wise Profit Margin:")
for i in range(len(stores)):
    print(stores[i], ":", store_profit_margin[i], "%")
```

```python
import numpy as np

sales_data = np.array([
    [200, 150, 180, 220],
    [250, 170, 200, 240],
    [180, 160, 190, 210],
    [300, 200, 220, 260]
])

cost_data = np.array([
    [80, 60, 50, 70],
    [90, 65, 55, 75],
    [85, 70, 60, 80],
    [95, 75, 65, 85]
])

categories = np.array([
    "Electronics",
    "Clothing",
    "Groceries",
    "Furniture"
])

profit_data = sales_data - cost_data

total_category_sales = np.sum(sales_data, axis=0)
total_category_profit = np.sum(profit_data, axis=0)

category_profit_margin = (total_category_profit / total_category_sales) * 100
category_profit_margin = np.round(category_profit_margin, 2)

best_category_index = np.argmax(category_profit_margin)

print("Best Category by Profit Margin:", categories[best_category_index])
print("Profit Margin:", category_profit_margin[best_category_index], "%")
```

Question 40: Low-Margin Transactions
Management considers a profit margin below 22% to be weak.

Find all store-category combinations where the profit margin is below 22%.

Display:

Store
Category
Sales
Cost
Profit
Profit margin
Use Boolean indexing.

```python
import numpy as np

sales_data = np.array([
    [200, 150, 180, 220],
    [250, 170, 200, 240],
    [180, 160, 190, 210],
    [300, 200, 220, 260]
])

cost_data = np.array([
    [80, 60, 50, 70],
    [90, 65, 55, 75],
    [85, 70, 60, 80],
    [95, 75, 65, 85]
])

stores = np.array([
    "Bengaluru",
    "Hyderabad",
    "Chennai",
    "Mumbai"
])

categories = np.array([
    "Electronics",
    "Clothing",
    "Groceries",
    "Furniture"
])

profit_data = sales_data - cost_data

profit_margin = (profit_data / sales_data) * 100
profit_margin = np.round(profit_margin, 2)

low_margin = profit_margin < 22

store_index, category_index = np.where(low_margin)

print("Low-Margin Transactions:")
for i in range(len(store_index)):
    s = store_index[i]
    c = category_index[i]

    print("Store:", stores[s])
    print("Category:", categories[c])
    print("Sales:", sales_data[s, c])
    print("Cost:", cost_data[s, c])
    print("Profit:", profit_data[s, c])
    print("Profit Margin:", profit_margin[s, c], "%")
    print("----------------------")
```
