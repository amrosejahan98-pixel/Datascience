```python
products = [
    {"name": "Laptop", "price": 55000, "stock": 5},
    {"name": "Mouse", "price": 600, "stock": 25},
    {"name": "Keyboard", "price": 1500, "stock": 0},
    {"name": "Monitor", "price": 18000, "stock": 8},
    {"name": "Printer", "price": 12000, "stock": 0},
    {"name": "SSD", "price": 6500, "stock": 12}
]

# Extract products currently in stock using filter()
available_products = filter(lambda product: product["stock"] > 0, products)

# Calculate inventory value using map()
inventory_values = list(map(
    lambda product: product["price"] * product["stock"],
    available_products
))

# Calculate total inventory value using reduce()
total_inventory_value = reduce(lambda x, y: x + y, inventory_values)

print("Inventory values:", inventory_values)
print("Total inventory value:", total_inventory_value)
```

```python
products = [
    {"name": "Laptop", "price": 55000, "category": "Electronics"},
    {"name": "Chair", "price": 4500, "category": "Furniture"},
    {"name": "Monitor", "price": 18000, "category": "Electronics"},
    {"name": "Desk", "price": 8500, "category": "Furniture"},
    {"name": "SSD", "price": 6500, "category": "Electronics"}
]

# Extract Electronics products using filter()
electronics_products = filter(
    lambda product: product["category"] == "Electronics",
    products
)

# Apply 20% discount using map()
discounted_prices = list(map(
    lambda product: product["price"] * 0.80,
    electronics_products
))

# Calculate total discounted value using reduce()
total_discounted_value = reduce(
    lambda x, y: x + y,
    discounted_prices
)

print("Discounted prices:", discounted_prices)
print("Total discounted value:", total_discounted_value)

Output:

Discounted prices: [44000.0, 14400.0, 5200.0]
Total discounted value: 63600.0

Explanation:

filter() extracts Electronics products:
Laptop, Monitor, SSD
map() applies a 20% discount (multiply by 0.80):
Laptop: 55000 × 0.80 = 44000
Monitor: 18000 × 0.80 = 14400
SSD: 6500 × 0.80 = 5200
reduce() adds discounted values:
44000 + 14400 + 5200 = 63600



```

```python
orders = [
    {"order_id": 1001, "amount": 1200, "status": "Delivered"},
    {"order_id": 1002, "amount": 3500, "status": "Pending"},
    {"order_id": 1003, "amount": 8500, "status": "Delivered"},
    {"order_id": 1004, "amount": 4200, "status": "Cancelled"},
    {"order_id": 1005, "amount": 7600, "status": "Delivered"},
    {"order_id": 1006, "amount": 5100, "status": "Pending"}
]

# Extract delivered orders using filter()
delivered_orders = list(filter(
    lambda order: order["status"] == "Delivered",
    orders
))

# Calculate total delivered revenue using reduce()
total_delivered_revenue = reduce(
    lambda total, order: total + order["amount"],
    delivered_orders,
    0
)

print("Delivered Orders:", delivered_orders)
print("Total Delivered Revenue:", total_delivered_revenue)
```

```python
transactions = [
    {"id": 1, "amount": 1200},
    {"id": 2, "amount": 5200},
    {"id": 3, "amount": 18000},
    {"id": 4, "amount": 2500},
    {"id": 5, "amount": 9600},
    {"id": 6, "amount": 7200}
]

# Extract transactions greater than 5000 using filter()
selected_transactions = filter(
    lambda transaction: transaction["amount"] > 5000,
    transactions
)

# Add 18% GST using map()
gst_transactions = list(map(
    lambda transaction: transaction["amount"] * 1.18,
    selected_transactions
))

# Calculate grand total using reduce()
grand_total = reduce(
    lambda x, y: x + y,
    gst_transactions
)

print("GST applied amounts:", gst_transactions)
print("Grand total after GST:", grand_total)
```

```python
sales_records = [
    {"salesperson": "Rahul", "region": "South", "sales": 45000, "active": True},
    {"salesperson": "Anil", "region": "North", "sales": 65000, "active": True},
    {"salesperson": "Sneha", "region": "South", "sales": 72000, "active": False},
    {"salesperson": "Kiran", "region": "South", "sales": 82000, "active": True},
    {"salesperson": "Meena", "region": "West", "sales": 55000, "active": True},
    {"salesperson": "Arjun", "region": "South", "sales": 98000, "active": True}
]

# Extract active South region salespeople using filter()
active_south_sales = filter(
    lambda person: person["region"] == "South" and person["active"],
    sales_records
)

# Calculate 5% incentive and create required list using map()
incentive_details = list(map(
    lambda person: {
        "salesperson": person["salesperson"],
        "sales": person["sales"],
        "incentive": person["sales"] * 0.05
    },
    active_south_sales
))

# Calculate total incentive using reduce()
total_incentive = reduce(
    lambda total, person: total + person["incentive"],
    incentive_details,
    0
)

print("Incentive Details:", incentive_details)
print("Total Incentive Payable:", total_incentive)

```
